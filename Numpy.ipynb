```python
a=np.array([[2,3,4],[3,4,5]],dtype=float)
print(a,a.shape)
print(type(a))
print(a.ndim)
print(a.size)
b=a.astype(float)
print(b)
```

```python
a=np.zeros((2,10),dtype=int,order='F')
print(a)
a=np.ones((10,5),dtype=int,order='F')
print(a)
a=np.full((10,3),5,dtype=int)
print(a)
a=np.arange(6)
print(a)
a=np.linspace(10,15,15,endpoint=True,dtype=int)
print(a,a.size)
a=np.eye(10,10,-3)
print(a)
a=np.random.rand(2,4)
print(a)
a=np.random.randn(2,4)
print(a,a.sum(),a.mean())
a=np.random.randint(1,10,(3,4))
print(a)
```

```python
a=np.array([[1,2,3,4],[3,4,5,6],[3,2,1,4]])
print(a,a.shape,a.size)
print(a[a>10])
```

```python
a=np.array([1,2,3,4])
b=np.array([1,2,3,4])
print(a+b)
print(a-b)
print(a*b)
print(a/2)
print(a>2)
print(np.sqrt(a))
print(np.square(a))
```

```python
a=np.array([1,2,3])
b=np.array([[1,2,3],[1,2,3],[1,2,3],[1,2,3]])
print(a.shape,b.shape)
print(a+b)
```

```python
a=np.array([[2,3,4,4],[3,5,3,5]])
print(a.reshape(2,2))
print(a.flatten(order='F'))
b=a.ravel()
print(b,b.shape)
print(a.T)
print(np.transpose(a))
print(a.shape)
print(np.expand_dims(a,axis=0))
print(np.squeeze(a,axis=None))
```

```python
a=np.array([1,2,9,3,4,5])
print(a.sum(),a.mean(),a.min(),a.max())
print(a.std(),a.var(),a.argmin(),a.argmax())
```

```python
a=np.array([[1,2],[2,1]])
b=np.array([[2,3,2],[3,2,2]])
print((a@b).shape)
print(np.matmul(a,b).shape)
print(np.dot(a,b).shape)
a=np.array([[1,2],[2,1]])
print(np.linalg.inv(a))
print(np.linalg.det(a))
print(np.linalg.eig(a))
```

```python
a=np.array([[1,2,3,4],[1,2,3,4]])
print(np.split(a,2))
print(np.concatenate((a,a)))
print(np.hsplit(a,2))
print(np.vsplit(a,2))
```
